```python
import pandas as pd
import numpy as np

# Wczytanie zbioru danych
data_set_path = "IHME_GBD_2019_SMOKING_AGE_1990_2019_PREV_Y2021M05D27.CSV"
df = pd.read_csv(data_set_path, delimiter=",")

# ===================================================================================
# Filtrowanie danych: Wybieramy tylko wskaźnik 'Prevalence' (odsetek palaczy),
# żeby nie mieszać ułamków z ogólną liczbą palaczy liczoną w milionach.
# ===================================================================================
df_prev = df[df["measure_name"] == "Prevalence"].copy()

print("===================================================================================")
print("PODGLĄD DANYCH: WSKAŹNIK 'PREVALENCE' (ODSETEK PALACZY W WIEKU 15-24 LAT)")
print(f"Liczba analizowanych wierszy: {len(df_prev)}")
print("===================================================================================\n")

# 2.1. Średnia arytmetyczna, wartość minimalna i maksymalna
max_val = np.max(df_prev["upper"])
min_val = np.min(df_prev["upper"])
mean_val = np.mean(df_prev["upper"])

print("--- 2.1. Średnia arytmetyczna (Mean) ---")
print(f"Max: {max_val:.4f}")
print(f"Min: {min_val:.4f}")
print(f"Średnia: {mean_val:.4f}")
print("Wniosek: Średnie górne oszacowanie liczby palących w badanej grupie to około 21%. "
      "Z kolei wartość maksymalna uświadamia, że są na świecie regiony, gdzie odsetek ten "
      "sięga nawet 69%.")

# 2.2. Mediana
median_val = np.median(df_prev["upper"])

print("\n--- 2.2. Mediana (Median) ---")
print(f"Mediana: {median_val:.4f}")
print("Wniosek: Mediana (19%) jest bliska średniej. Pokazuje to, że w 'typowym' kraju "
      "pali blisko co piąta młoda osoba, a nasze wyniki nie są już sztucznie zawyżane "
      "przez żadne wartości skrajne.")

# 2.3. Odchylenie standardowe
std_dev = np.std(df_prev["upper"])

print("\n--- 2.3. Odchylenie standardowe (Std) ---")
print(f"Odchylenie standardowe: {std_dev:.4f}")
print("Wniosek: Wynik na poziomie 0.14 (14 punktów procentowych) oznacza, że w zależności "
      "od kraju i płci, odsetek palaczy dość mocno waha się wokół naszej średniej.")

# 2.4. Wariancja
variance = np.var(df_prev["upper"])

print("\n--- 2.4. Wariancja (Var) ---")
print(f"Wariancja: {variance:.4f}")
print("Wniosek: Jest to kwadrat odchylenia standardowego, po prostu inna matematyczna "
      "miara rozproszenia tych oszacowań.")

# 2.5. Korelacja
correlation_np = np.corrcoef(df_prev["upper"], df_prev["lower"])[0, 1]

print("\n--- 2.5. Korelacja ('upper' vs 'lower') ---")
print(f"Korelacja (np.corrcoef): {correlation_np:.4f}")
print("Wniosek: Bardzo wysoka korelacja (~0.97) oznacza, że dolna i górna granica "
      "szacunków są ze sobą silnie powiązane – rosną i maleją praktycznie równocześnie.")

# 2.6. Kowariancja i ręczne wyliczenie korelacji
covariance = np.cov(df_prev["upper"], df_prev["lower"])
covariance_val = covariance[0, 1]
corelation_manual = covariance[0, 1] / (np.sqrt(covariance[0, 0]) * np.sqrt(covariance[1, 1]))

print("\n--- 2.6. Kowariancja ('upper' vs 'lower') ---")
print("Macierz kowariancji:")
print(covariance)
print(f"Kowariancja [0, 1]: {covariance_val:.6f}")
print(f"Korelacja (ze wzoru): {corelation_manual:.4f}")
print("Wniosek: Dodatnia kowariancja po prostu matematycznie potwierdza to, co widzieliśmy "
      "przed chwilą – obie granice przedziału ufności przesuwają się w tę samą stronę.")
```
