```python
# Import niezbędnych bibliotek i ustawienie wykresów
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline

# Tworzenie ramki danych ze słownika
slownik = {
    'Kraj': ['Polska', 'Niemcy', 'Czechy'],
    'Odsetek_palacych': [15.2, 12.1, 14.5],
    'Rok': [2019, 2019, 2019]
}
df_ze_slownika = pd.DataFrame(slownik)
print("--- Ramka ze słownika ---")
display(df_ze_slownika)

# Zachowanie ramki danych pobranych z pliku CSV 
nazwa_pliku = 'IHME_GBD_2019_SMOKING_AGE_1990_2019_PREV_Y2021M05D27.CSV'
df = pd.read_csv(nazwa_pliku)
print("\n--- Pomyślnie wczytano dane z pliku CSV ---")

# Tworzenie ramki danych z listy list
lista_list = [
    ['Francja', 13.0, 2019],
    ['Hiszpania', 16.2, 2019],
    ['Włochy', 14.8, 2019]
]
df_z_listy = pd.DataFrame(lista_list, columns=['Kraj', 'Odsetek_palacych', 'Rok'])
print("\n--- Ramka z listy list ---")
display(df_z_listy)

# Transponowanie (zamiana kolumn i wierszy)
df_transponowana = df_ze_slownika.T
print("\n--- Ramka transponowana ---")
display(df_transponowana)

# Wyświetlić pierwsze 10 wierszy
print("--- Pierwsze 10 wierszy ---")
display(df.head(10))

# Wyświetlić ostatnie 10 wierszy
print("\n--- Ostatnie 10 wierszy ---")
display(df.tail(10))

# Informacja o ramce danych
print("\n--- Informacje o ramce danych ---")
df.info()

# Ile wierszy i kolumn
wiersze, kolumny = df.shape
print(f"\nRamka ma {wiersze} wierszy oraz {kolumny} kolumn.")

# Statystyki kolumn liczbowych
print("\n--- Statystyki kolumn liczbowych ---")
display(df.describe())

# Statystyki kolumn kategoryzowanych
print("\n--- Statystyki kolumn kategoryzowanych ---")
display(df.describe(include=['object']))

# Usunięcie brakujących wartości
df = df.dropna()
print(f"\nRozmiar ramki po usunięciu brakujących wartości: {df.shape}")

# Wybór wierszy i kolumn po nazwach i indeksach
display(df.loc[0:5, ['location_name', 'val']]) 
display(df.iloc[0:5, [0, 2]])                  

# Wybór wierszy pod warunkiem określonej wartości
display(df[df['year_id'] == '2019'].head())

# Wybór wierszy przy spełnieniu kilku warunków
display(df[(df['year_id'] == '2019') & (df['val'] > 0.1)].head())

# Wiersze, które zawierają w kolumnie określone słowo
display(df[df['location_name'].str.contains('Poland', na=False)].head())

# Wiersze, które nie zawierają określonego słowa
display(df[~df['location_name'].str.contains('Poland', na=False)].head())

# Utworzenie nowej kolumny na podstawie istniejącej
df['val_procent'] = df['val'] * 100

# Usunięcie kolumny
df = df.drop(columns=['val_procent'])

# Zmiana nazwy kolumny
df = df.rename(columns={'val': 'Wartosc'})

# Zachowanie ramki jako plik csv
df.to_csv('wariant5_przetworzone.csv', index=False)

# Średnia, maksymalna i minimalna wartość z jednej kolumny
print("Średnia:", df['Wartosc'].mean())
print("Maksimum:", df['Wartosc'].max())
print("Minimum:", df['Wartosc'].min())

# Wyświetlenie liczby wierszy
print("Liczba wierszy:", len(df))

# Wartości unikatowe w kolumnie
print("Unikalne lata:", df['year_id'].unique())

# Liczby rekordów odpowiadających wartościom
print(df['location_name'].value_counts().head())

# Sortowanie wierszy malejąco i rosnąco
display(df.sort_values(by='Wartosc', ascending=False).head())
display(df.sort_values(by='Wartosc', ascending=True).head())

# 10 największych i najmniejszych wartości
display(df.nlargest(10, 'Wartosc'))
display(df.nsmallest(10, 'Wartosc'))

# 10 największych wartości pod warunkiem określonej wartości innej kolumny
dane_2019 = df[df['year_id'] == '2019']
display(dane_2019.nlargest(10, 'Wartosc'))
# Grupowanie (MultiIndex) i uśrednienie wartości wszystkich kolumn
kolumny_liczbowe = df.select_dtypes(include=['number']).columns
df_grupowane = df.groupby(['location_name', 'year_id'])[kolumny_liczbowe].mean()
display(df_grupowane.head())

# Grupowanie - średnia, liczba, mediana
display(df.groupby('location_name').agg({
    'Wartosc': 'mean',
    'year_id': 'count',
    'upper': 'median' 
}).head())

# Nazwy kolumn indeksu złożonego
print("Nazwy w MultiIndex:", df_grupowane.index.names)

# Sortowanie kolumny indeksu złożonego
display(df_grupowane.sort_index(level='location_name', ascending=False).head())

# Tabela przestawna (pivot table)
tabela_przestawna = pd.pivot_table(df, values='Wartosc', index=['location_name'], columns=['year_id'], aggfunc='mean')

# Wyświetlenie indeksów i kolumn tabeli przestawnej
print("Indeksy tabeli:", tabela_przestawna.index)
print("Kolumny tabeli:", tabela_przestawna.columns)

# Indeks złożony tabeli przestawnej
tabela_zlozona = pd.pivot_table(df, values='Wartosc', index=['location_name', 'sex_name'], columns=['year_id'], aggfunc='mean')
display(tabela_zlozona.head())

# Wykres na podstawie tabeli przestawnej
tabela_przestawna.head(5).T.plot(figsize=(10, 6), title="Wartość wskaźnika w czasie dla 5 krajów")
plt.show()


# Narysować histogram na podstawie wartości kolumny (tylko dla Prevalence)
df_prevalence = df[df['measure_name'] == 'Prevalence']
df_prevalence['Wartosc'].plot(kind='hist', bins=20, title="Histogram: Rozkład Wartości (Prevalence)")
plt.show()

# Metody merge i concat (łączenie)
df_czesc1 = df.head(100)
df_czesc2 = df.tail(100)
df_concat = pd.concat([df_czesc1, df_czesc2])

df_slownik = pd.DataFrame({'location_name': ['Poland', 'Germany'], 'Kod_kraju': ['PL', 'DE']})
df_merge = pd.merge(df, df_slownik, on='location_name', how='inner')
display(df_merge.head())

# Nowe kolumny - operacje matematyczne
df['Roznica'] = df['upper'] - df['lower']

# Nowe kolumny - funkcja lambda
df['Kategoria'] = df['Wartosc'].apply(lambda x: 'Wysoka' if x > 0.2 else 'Niska')

# Praca z dużymi plikami (chunksize)
print("Przetwarzanie pliku partiami (chunksize):")
for paczka in pd.read_csv(nazwa_pliku, chunksize=50000):
    print(f"Rozmiar pobranej paczki: {paczka.shape}")
    break
```
